```python
def index_to_datetime(panda_dataframe):

    '''Merging the date and the daytime data to create a new index for the dataframe'''

    time = panda_dataframe['Time(hh:mm:ss)'].values.astype('str')
    date = panda_dataframe.index.values.astype('str')
    date_time = np.empty(time.size).astype('str')

    for i in range(date.size):
        try :
            date_time[i] = (pd.to_datetime(str(date[i]) + ' ' + str(time[i]), format='%d:%m:%Y %H:%M:%S'))

        except ValueError :
            date_time[i] = np.nan

    final_dataframe = panda_dataframe
    final_dataframe.index = date_time
    del final_dataframe['Time(hh:mm:ss)']

    return final_dataframe


def datetime_to_timestamp(date_time):

    '''Converting a panda.Datetime object to a timestamp float object in seconds'''

    epoch = datetime(1970, 1, 1)
    dt = np.datetime64(date_time)
    dtr = dt.astype('M8[ms]').astype('O')
    dtf = (dtr - epoch) / timedelta(seconds=1)
    return dtf

```

```python
site_name = 'Lille'
year1 = '2015'
month1 = '1'
day1 = '01'
year2 = '2015'
month2 = '12'
day2 = '31'
AODversion = '15'
AVG = '20'  # All points: AVG=10, Daily average: AVG=20'''

csv_url = 'https://aeronet.gsfc.nasa.gov/cgi-bin/print_web_data_v3?site=' + site_name + \
          '&year=' + year1 + '&month=' + month1 + '&day=' + day1 + \
          '&year2=' + year2 + '&month2=' + month2 + '&day2=' + day2 + \
          '&AOD' + AODversion + '=1&AVG=' + AVG
print(csv_url)


csv_data = pd.read_csv(csv_url, skiprows=7, error_bad_lines=False, parse_dates=['Date(dd-mm-yyyy)'], dayfirst=True,
                       index_col='Date(dd-mm-yyyy)')

print('csv ok')
csv_data = csv_data.drop(csv_data.index[len(csv_data) - 1])
csv_data = index_to_datetime(csv_data)
aot440 = csv_data.AOD_440nm.values
angstrom = csv_data['440-675_Angstrom_Exponent'].values
aot550 = ((550 / 440) ** (-angstrom)) * aot440
lat = csv_data['Site_Latitude(Degrees)'].values[0]
lon = csv_data['Site_Longitude(Degrees)'].values[0]

taup550 = xarray.Dataset({'aeronet_taup550': (('time'), aot550)},
                     coords={'time': csv_data.index.values.astype('datetime64[ns]'),
                             'lat': lat,
                             'lon': lon})
taup550
```
